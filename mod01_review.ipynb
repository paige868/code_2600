### Create a subset of Credit that only contains the columns Age, Region, and Income. Call it "Subset."

```python
import csv

Subset = []
cats = ["Age", "Region", "Income"]

with open('data/Credit.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        picks = {key: row[key] for key in cats}
        Subset.append(picks)
       
```

### From this subset, create two partitions: 
- One that contains records for Ages below 50, and one that contains Ages 50+
- Call them "Younger" and "Older"

```python

import csv
import statistics

Ages = []
Younger =[]
Older = []

with open('data/Credit.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        Ages.append(int(row['Age']))

for age in Ages:
    if age < 50:
            Younger.append(age)
    else:
         Older.append(age)
         
       
```

Often, banks will use a variable called _utilization_ to assess credit risk.
Utilization is the ratio of balance to limit (i.e., what % of the credit limit is being used?)
### Add a feature to the Credit dataset to measure utilization

```python
import csv
import statistics 
import os

data_with_utilization = []

with open('data/Credit.csv', 'r') as file:
    reader = csv.DictReader(file)
    

    for row in reader:
        balance = float(row["Balance"])
        limit = float(row["Limit"])
        row['Utilization'] = balance/limit
        data_with_utilization.append(row)




```

### What is the maximum utilization value in this set? 

```python
utilization_max = 0
for row in data_with_utilization:
    ut = row["Utilization"]
   
    if ut > utilization_max:
        utilization_max = ut

```
